# set-up the project environment for formula1_incr project
1. Create external location databricks-course-exter-dl1-formula1-incr


###Access cloud storage

```sql
%fs ls 'abfss://formula1-incr@databrickscourseexterdl1.dfs.core.windows.net/landing'
```

### create external location

```sql
-- Create a location accessed using the abfss_remote_cred credential
CREATE EXTERNAL LOCATION IF NOT EXISTS `databricks_course_exter_dl1_formula1_incr` URL 'abfss://formula1-incr@databrickscourseexterdl1.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-course-sc`)
    COMMENT 'External loc for formula1_incr container';
```

```sql
SHOW CATALOGS ;
```

### Create catalog formula1_incr

```sql
create catalog if not exists formula1_incr
    managed location 'abfss://formula1-incr@databrickscourseexterdl1.dfs.core.windows.net/' ;
```

### create schemas landing, bronze, silver, gold

```sql
CREATE SCHEMA IF NOT EXISTS formula1_incr.landing;
CREATE SCHEMA IF NOT EXISTS formula1_incr.bronze
    MANAGED LOCATION 'abfss://formula1-incr@databrickscourseexterdl1.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1_incr.silver
    MANAGED LOCATION 'abfss://formula1-incr@databrickscourseexterdl1.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS formula1_incr.gold
    MANAGED LOCATION 'abfss://formula1-incr@databrickscourseexterdl1.dfs.core.windows.net/gold';
```

```sql
use catalog formula1_incr;
```
